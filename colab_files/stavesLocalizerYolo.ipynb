```python
def visualize_results(test_results, result_show_limit=20):
  # Collect plotted images from results
  grid_imgs = []
  for i, result in enumerate(test_results):
      if i >= result_show_limit:
          break
      # result.plot() returns an image (NumPy array) with predictions overlaid
      img = result.plot()
      grid_imgs.append(img)

  # Determine grid dimensions (e.g., 4 columns; adjust rows accordingly)
  cols = 4
  rows = int(np.ceil(len(grid_imgs) / cols))

  # Create a figure to display the grid
  fig, axs = plt.subplots(rows, cols, figsize=(cols * 4, rows * 4))

  for idx, ax in enumerate(axs.flat):
      if idx < len(grid_imgs):
          # Convert BGR (if using OpenCV) to RGB if needed
          # img = cv2.cvtColor(grid_imgs[idx], cv2.COLOR_BGR2RGB)
          ax.imshow(grid_imgs[idx])
          ax.set_title(f"Result {idx+1}")
      ax.axis('off')

#conf thresholding at 80
def crop_and_save_images_from_results(test_results=None, output_folder=None, conf_thresh=0.8, class_id=0):
    cropped_images_folder = os.path.join(output_folder, "cropped_images")
    labels_folder = os.path.join(output_folder, "labels")
    os.makedirs(cropped_images_folder, exist_ok=True)
    os.makedirs(labels_folder, exist_ok=True)

    if test_results:
        for a_result in test_results:
            img = a_result.orig_img  #
            h, w, _ = img.shape

            img_path = a_result.path
            img_name = os.path.splitext(os.path.basename(img_path))[0]

            box_coords = a_result.boxes.xyxy.tolist()
            conf_scores = a_result.boxes.conf.tolist()

            yolo_lines = []

            for i, (a_conf, box) in enumerate(zip(conf_scores, box_coords)):
                if a_conf >= conf_thresh:
                    x1, y1, x2, y2 = map(int, box)

                    #img save
                    cropped = img[y1:y2, x1:x2]
                    crop_filename = f"{img_name}.jpeg"
                    cv2.imwrite(os.path.join(cropped_images_folder, crop_filename), cropped)

                    #label save
                    x_center = ((x1 + x2) / 2) / w
                    y_center = ((y1 + y2) / 2) / h
                    box_width = (x2 - x1) / w
                    box_height = (y2 - y1) / h
                    yolo_line = f"{class_id} {x_center:.6f} {y_center:.6f} {box_width:.6f} {box_height:.6f}"
                    yolo_lines.append(yolo_line)

            #create empty label files for no detections
            label_file_path = os.path.join(labels_folder, f"{img_name}.txt")
            with open(label_file_path, 'w') as f:
                for line in yolo_lines:
                    f.write(line + "\n")
```

```python
weights_path = "/content/drive/MyDrive/Stave_Project/Developers/modelWeights/staveLocalizerTrainResultMar19/detect/train/weights/best.pt"
input_folder = "/content/drive/MyDrive/Stave_Project/Developers/staves200ForLocalizer"
output_folder = "/content/drive/MyDrive/Stave_Project/Developers/staveProjectImgsCropped"
```

```python
model = ultralytics.YOLO(weights_path)
test_results = []
for result in model.predict(input_folder, half=True, amp=True, max_det=1):
  test_results.append(result)
```

```python
crop_and_save_images_from_results(test_results, output_folder)
```

```python
visualize_results(test_results, result_show_limit=50)
```
